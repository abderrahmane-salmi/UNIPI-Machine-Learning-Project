```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import train_test_split, GridSearchCV, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import make_scorer
from sklearn.compose import TransformedTargetRegressor
```

```python
from sklearn.metrics import mean_squared_error

def plot_learning_curves(grid_search, X_train, y_train, X_test, y_test, epochs=300):
    """
    Plots MSE and MEE (Accuracy) learning curves for the best model 
    found by GridSearchCV in nn-2.ipynb.
    """
    # 1. Clean parameters (remove 'regressor__' prefix)
    best_params = grid_search.best_params_
    params = {k.split('__')[-1]: v for k, v in best_params.items()}
    
    # 2. Handle 'lbfgs' (it doesn't support incremental training for curves)
    if params.get('solver') == 'lbfgs':
        print("INFO: Solver 'lbfgs' doesn't support partial_fit. Using 'adam' for curves.")
        params['solver'] = 'adam'
        if 'learning_rate_init' not in params: params['learning_rate_init'] = 0.001
    
    # Remove non-training parameters
    params.pop('max_iter', None)
    params.pop('early_stopping', None)

    # 3. Setup Scaling for y (important for MLPRegressor)
    # We use the transformer found in your best estimator
    y_scaler = grid_search.best_estimator_.transformer_
    y_train_scaled = y_scaler.transform(y_train)

    # 4. Initialize model
    model = MLPRegressor(**params, random_state=42)
    
    train_mse, test_mse = [], []
    train_mee, test_mee = [], []

    def calc_mee(y_true, y_pred):
        # Mean Euclidean Error (The 'accuracy' for ML Cup)
        return np.mean(np.sqrt(np.sum((y_true - y_pred)**2, axis=1)))

    # 5. Training Loop
    for i in range(epochs):
        model.partial_fit(X_train, y_train_scaled)
        
        # Get predictions and unscale them
        p_train_scaled = model.predict(X_train)
        p_test_scaled = model.predict(X_test)
        
        # Reshape for inverse transform (handles multi-output)
        y_dim = y_train.shape[1]
        p_train = y_scaler.inverse_transform(p_train_scaled.reshape(-1, y_dim))
        p_test = y_scaler.inverse_transform(p_test_scaled.reshape(-1, y_dim))
        
        # Calculate Metrics
        train_mse.append(mean_squared_error(y_train, p_train))
        test_mse.append(mean_squared_error(y_test, p_test))
        train_mee.append(calc_mee(y_train, p_train))
        test_mee.append(calc_mee(y_test, p_test))

    # 6. Plotting
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))
    
    # MSE Plot
    ax1.plot(train_mse, label='Train MSE', color='blue')
    ax1.plot(test_mse, label='Test MSE', color='red')
    ax1.set_title('Learning Curve: MSE (Loss)')
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Mean Squared Error')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # Accuracy (MEE) Plot
    ax2.plot(train_mee, label='Train MEE', color='blue')
    ax2.plot(test_mee, label='Test MEE', color='red')
    ax2.set_title('Learning Curve: MEE (Accuracy)')
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('Mean Euclidean Error')
    ax2.legend()
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
```

```python
import sys
from pathlib import Path

# Get current notebook directory
current_dir = Path.cwd()

# Move up until we find 'src'
for parent in current_dir.parents:
    if parent.name == "src":
        sys.path.append(str(parent))
        break
else:
    raise RuntimeError("Could not find 'src' directory")

from common import utils
```

```python
ids, X, y = utils.load_cup_data(train=True)
```

```python
X_dev, X_int_test, y_dev, y_int_test = utils.split_data(X, y, random_state=42)
```

```python
x_scaler = StandardScaler()
y_scaler = StandardScaler()
X_dev_scaled = x_scaler.fit_transform(X_dev)
X_test_scaled = x_scaler.transform(X_int_test) # Transform test using dev params
```

```python
model = TransformedTargetRegressor(
    regressor=MLPRegressor(random_state=42, max_iter=2000, early_stopping=True),
    transformer=StandardScaler()
)
```

```python
# Define Custom MEE Scorer (Greater is better for sklearn, so return negative MEE)
def mean_euclidean_error(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    # Euclidean distance per sample: sqrt(sum of squared differences)
    errors = np.sqrt(np.sum((y_true - y_pred) ** 2, axis=1))
    return np.mean(errors)

# greater_is_better=False makes it a loss function (sklearn will minimize it)
mee_scorer = make_scorer(mean_euclidean_error, greater_is_better=False)
```

```python
param_grid = {
    'regressor__hidden_layer_sizes': [(50,), (20, 20), (40, 30), (50, 20), (70, 40),(100, 50), (30, 20, 10)],
    'regressor__activation': ['relu', 'tanh'],
    'regressor__solver': ['adam', 'sgd', 'lbfgs'],
    'regressor__alpha': [0.0001, 0.001, 0.01, 0.1], # L2 Regularization (Required!)
    'regressor__learning_rate_init': [0.001, 0.01],
    'regressor__momentum': [0.5, 0.7, 0.9] # Relevant for sgd
}

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

grid_search = GridSearchCV(
    model,
    param_grid,
    cv=kfold,
    scoring=mee_scorer,
    n_jobs=-1,
    return_train_score=False, # Save memory
    verbose=1
)

print("Starting Grid Search...")
# Note: We pass y_dev (raw), the wrapper handles scaling!
grid_search.fit(X_dev_scaled, y_dev)
```

```python
# 1. Convert results to DataFrame
results_df = pd.DataFrame(grid_search.cv_results_)

# 2. Sort by rank (Rank 1 is the best/lowest error)
top_5_results = results_df.nsmallest(10, 'rank_test_score')

print(f"{'Rank':<5} | {'MEE (Positive)':<15}")
print("-" * 50)

for _, row in top_5_results.iterrows():
    raw_score = row['mean_test_score']
    # Flip sign if your scorer returns negative values
    positive_mee = -raw_score if raw_score < 0 else raw_score
    rank = row['rank_test_score']

    print(f"{rank:<5} | {positive_mee:<15.4f}")
    print(f"Params: {row['params']}\n")
```

```python
# Assuming you have X_dev_scaled, y_dev, X_test_scaled, and y_int_test
plot_learning_curves(
    grid_search, 
    X_dev_scaled, y_dev, 
    X_test_scaled, y_int_test, 
    epochs=300
)
```
